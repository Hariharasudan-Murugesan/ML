```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder,StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
rf=RandomForestRegressor()
dt=DecisionTreeRegressor()
lr=LinearRegression()
sc=StandardScaler()
le=LabelEncoder()
class housing_data:
    def get_in(self):
        n=int(input("Enter how many dataset you Want to Enter"))
        Price=[]
        Area=[]
        Bedrooms=[]
        Bathrooms=[]
        Stories=[]
        Mainroad=[]
        Guestroom=[]
        Basement=[]
        Hotwaterheating=[]
        Airconditioning=[]
        Parking=[]
        Prefarea=[]
        Furnishingstatus=[]
        
        for i in range(n):
            price=int(input("Enter the price"))
            Price.append(price)
            area=int(input("Enter area "))
            Area.append(area)
            bedrooms=int(input("Enter no of bedrooms"))
            Bedrooms.append(bedrooms)
            bathrooms=int(input("Enter no of bathrooms"))
            Bathrooms.append(bathrooms)
            stories=int(input("Enter no of stories"))
            Stories.append(stories)
            mainroad=input("Enter its on mainroad yes or no ").lower()
            Mainroad.append(mainroad)
            guestroom=input("Enter its have guestroom yes or no ").lower()
            Guestroom.append(guestroom)
            basement=input("Enter its have basement yes or no ").lower()
            Basement.append(basement)
            hotwaterheating=input("Enter its have hotwaterheating yes or no ").lower()
            Hotwaterheating.append(hotwaterheating)
            airconditioning=input("Enter its have airconditioning yes or no ").lower()
            Airconditioning.append(airconditioning)
            parking=int(input("Enter NO of parking"))
            Parking.append(parking)
            prefarea=input("Enter its have prefarea yes or no ").lower()
            Prefarea.append(prefarea)
            furnishingstatus=input("Enter furnishing status furnished,semi-furnished,unfurnished").lower()
            Furnishingstatus.append(furnishingstatus)
        df={"price":Price,"area":Area,"bedrooms":Bedrooms,"bathrooms":Bathrooms,"stories":Stories,
            "mainroad":Mainroad,"guestroom":Guestroom,"basement":Basement,
            "hotwaterheating":Hotwaterheating,"airconditioning":Airconditioning,"parking":Parking,
            "prefarea":Prefarea,"furnishingstatus":Furnishingstatus}
        
        df=pd.DataFrame(df)
        self.df=df
        
class prepros(housing_data):
    def labelen(self):
        self.df["mainroad"]=le.fit_transform(self.df["mainroad"])
        self.df["guestroom"]=le.fit_transform(self.df["guestroom"])
        self.df["basement"]=le.fit_transform(self.df["basement"])
        self.df["hotwaterheating"]=le.fit_transform(self.df["hotwaterheating"])
        self.df["airconditioning"]=le.fit_transform(self.df["airconditioning"])
        self.df["prefarea"]=le.fit_transform(self.df["prefarea"])
        self.df["furnishingstatus"]=le.fit_transform(self.df["furnishingstatus"])
            
    def TrainTestSplit(self):
        x=self.df.drop(columns=["price"])
        y=self.df["price"]
        x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
        self.x_train=x_train
        self.x_test=x_test
        self.y_train=y_train
        self.y_test=y_test
        
class LinearReg(prepros):
    def model(self):
        self.x_train=sc.fit_transform(self.x_train)
        self.x_test=sc.transform(self.x_test)
        lr.fit(self.x_train,self.y_train)
        self.y_predict=lr.predict(self.x_test)
    def Accuracy(self):
        accuracy=r2_score(self.y_test,self.y_predict)
        print("Accuracy:",accuracy*100)

class Decsion_tree(prepros):
    def model(self):
        dt.fit(self.x_train,self.y_train)
        self.y_predict=dt.predict(self.x_test)
    def Accuracy(self):
        accuracy=r2_score(self.y_test,self.y_predict)
        print("Accuracy:",accuracy*100)

class RandomFor(prepros):
    def model(self):
        rf.fit(self.x_train,self.y_train)
        self.y_predict=rf.predict(self.x_test)
    def Accuracy(self):
        accuracy=r2_score(self.y_test,self.y_predict)
        print("Accuracy:",accuracy*100)
class Menu:
    def selection(self):
        select=int(input("Select the Model \n 1.LinearRegression  \n 2.DecisionTreeRegressor \n 3.RandomForestRegressor"))
        if select==1:
            return(LinearReg())
        elif select==2:
            return(Decsion_tree())
        elif select==3:
            return(RandomFor())
        else:
            print("input Invalid")

menu=Menu()
h=menu.selection()
h.get_in()
h.labelen()
h.TrainTestSplit()
h.model()
h.Accuracy()
```
