```python
class lib:
    def libraries(self):
        import pandas as pd
        from sklearn.preprocessing import LabelEncoder
        from sklearn.model_selection import train_test_split
        from sklearn.preprocessing import StandardScaler
        from sklearn.linear_model import LogisticRegression
        from sklearn.svm import SVC
        from sklearn.tree import DecisionTreeClassifier
        from sklearn.ensemble import RandomForestClassifier
        from sklearn.metrics import accuracy_score
        self.rf=RandomForestClassifier(n_estimators=100,random_state=42)
        self.dt=DecisionTreeClassifier(random_state=42)
        self.AS=accuracy_score
        self.lr=LogisticRegression(max_iter=500)
        self.model=SVC()
        self.sc=StandardScaler()
        self.TTS=train_test_split
        self.le=LabelEncoder()
        self.pd=pd
class datasets(lib):
    
    def read(self):
        self.df=self.pd.read_csv("https://raw.githubusercontent.com/YBIFoundation/Dataset/main/TelecomCustomerChurn.csv")
        
    def drp(self):
        self.df=self.df.drop("customerID",axis=1)
        
    def conv(self):
        self.df["TotalCharges"] = self.pd.to_numeric(self.df["TotalCharges"],errors="coerce")     
        
    def mis_val(self):
        self.df["TotalCharges"]=self.df["TotalCharges"].fillna(self.df["TotalCharges"].mean())
        
    def label(self):
        for col in self.df.select_dtypes(include="object").columns:
            if col != "customerID":
                self.df[col] = self.le.fit_transform(self.df[col])
        

    def tts(self):
        x=self.df.drop("Churn",axis=1)
        y=self.df["Churn"]
        self.x_train,self.x_test,self.y_train,self.y_test=self.TTS(x,y,test_size=0.2,random_state=42)
        
    def SS(self):
        self.x_train=self.sc.fit_transform(self.x_train)
        self.x_test=self.sc.transform(self.x_test)
        
class Logistic_Regression(datasets):
    
    def mode(self):
        self.lr.fit(self.x_train,self.y_train)
        self.y_pred=self.lr.predict(self.x_test)

    def prd(self):
        accuracy=self.AS(self.y_test,self.y_pred)
        print(f"Accuarcy :{accuracy*100}")
        
class Svm(datasets):
    
    def mode(self):
        self.model.fit(self.x_train,self.y_train)
        self.y_pred=self.model.predict(self.x_test)

    def prd(self):
        accuracy=self.AS(self.y_test,self.y_pred)
        print(f"Accuarcy :{accuracy*100}")

class DecisionTreeModel(datasets):
    
    def mode(self):
        self.dt.fit(self.x_train,self.y_train)
        self.y_pred=self.dt.predict(self.x_test)
        
    def prd(self):
        accuracy=self.AS(self.y_test,self.y_pred)
        print(f"Accuarcy :{accuracy*100}")
    
class RandomForestModel(datasets):
    def mode(self):
        self.rf.fit(self.x_train,self.y_train)
        self.y_pred=self.rf.predict(self.x_test)
        
    def prd(self):
        accuracy=self.AS(self.y_test,self.y_pred)
        print(f"Accuarcy :{accuracy*100}")

class option:
    def opt(self):
        x=int(input("Select the Model \n 1.Logistic Regression \n 2.SVM  \n 3.Decision Tree \n 4.Random Forest\n "))
        if x==1:
            return Logistic_Regression()
        elif x==2:
            return Svm()
        elif x==3:
            return DecisionTreeModel()
        elif x==4:
            return RandomForestModel()
        else:
            print("Invalid input")
            
        
        
menu=option()  
d=menu.opt()
d.libraries()
d.read()
d.drp()
d.conv()
d.mis_val()
d.label()
d.tts()
d.SS()
d.mode()
d.prd()
```
